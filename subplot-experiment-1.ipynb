```python
# EDA Charts Using Subplots and Random Data

import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

# Make random values reproducible
np.random.seed(42)

# Create random dataset
number_of_records = 100

df = pd.DataFrame({
    "Study_Hours": np.random.randint(1, 11, number_of_records),
    "Attendance": np.random.randint(50, 101, number_of_records),
    "Marks": np.random.randint(30, 101, number_of_records),
    "Department": np.random.choice(
        ["CSE", "ECE", "ME", "CE"],
        number_of_records
    )
})

print(df.head())


# Create 2 rows and 3 columns of subplots
fig, axes = plt.subplots(2, 3, figsize=(18, 10))

fig.suptitle(
    "EDA Charts Using Random Student Data",
    fontsize=20,
    fontweight="bold"
)


# 1. Line Chart
axes[0, 0].plot(
    df.index,
    df["Marks"],
    color="blue",
    linewidth=1.5
)

axes[0, 0].set_title("Line Chart of Marks")
axes[0, 0].set_xlabel("Student Number")
axes[0, 0].set_ylabel("Marks")


# 2. Bar Chart
department_counts = df["Department"].value_counts()

axes[0, 1].bar(
    department_counts.index,
    department_counts.values,
    color=["blue", "orange", "green", "red"]
)

axes[0, 1].set_title("Students by Department")
axes[0, 1].set_xlabel("Department")
axes[0, 1].set_ylabel("Number of Students")


# 3. Pie Chart
axes[0, 2].pie(
    department_counts.values,
    labels=department_counts.index,
    autopct="%1.1f%%",
    startangle=90,
    colors=sns.color_palette("pastel")
)

axes[0, 2].set_title("Department Distribution")


# 4. Scatter Plot
axes[1, 0].scatter(
    df["Study_Hours"],
    df["Marks"],
    color="green",
    alpha=0.7
)

axes[1, 0].set_title("Study Hours vs Marks")
axes[1, 0].set_xlabel("Study Hours")
axes[1, 0].set_ylabel("Marks")


# 5. Histogram
axes[1, 1].hist(
    df["Marks"],
    bins=10,
    color="orange",
    edgecolor="black"
)

axes[1, 1].set_title("Distribution of Marks")
axes[1, 1].set_xlabel("Marks")
axes[1, 1].set_ylabel("Frequency")


# 6. Box Plot
axes[1, 2].boxplot(
    df["Attendance"],
    patch_artist=True,
    boxprops={"facecolor": "lightblue"}
)

axes[1, 2].set_title("Attendance Box Plot")
axes[1, 2].set_ylabel("Attendance Percentage")
axes[1, 2].set_xticks([1])
axes[1, 2].set_xticklabels(["Attendance"])


# Adjust spacing between subplots
plt.tight_layout(rect=[0, 0, 1, 0.95])

# Display all charts
plt.show()
```
